**Paso 1 y 2 — Configurar credenciales y descargar el dataset.**

```python
# Instala la librería kaggle (si no está instalada ya)
!pip install kaggle --quiet

import os

# Carga el token desde los Secretos de Colab
try:
    from google.colab import userdata
    os.environ["OPEN_KAGGLE"] = userdata.get("OPEN_KAGGLE")
    print("Token de Kaggle cargado")
except ImportError:
    print("Esta celda está diseñada para ejecutarse en Google Colab con el Secreto configurado")

# Completa el identificador del dataset y descárgalo en datos_drug/
dataset_id = "prathamtripathi/drug-classification"  # <- completa aquí

!kaggle datasets download -d {dataset_id} -p datos_drug/ --unzip
```

**Paso 3 — Cargar el archivo descargado en un DataFrame.**

```python
import glob
import pandas as pd

archivos_csv = glob.glob("datos_drug/*.csv")
print(archivos_csv[0])

# Completa: carga el primer archivo de archivos_csv en un DataFrame llamado df_drug
df_drug = pd.read_csv(archivos_csv[0])
print(df_drug)
```

**Paso 4 — Verificar el resultado.**

```python
# Completa: imprime shape, columns, dtypes y valores faltantes de df_drug,
# siguiendo el mismo patrón usado en los Bloques 1 y 2.
print(df_drug.shape)    #Para obtener las dimensiones
print(df_drug.columns)  #Para obtener las columnas
print(df_drug.dtypes)   #Para obtener los tipos de datos
print("Valores faltantes por columna: \n",  df_drug.isna().sum(), sep="") #Para obtener los valores faltantes
```
